```python
with open("bank_data.csv",newline="",encoding="utf-8") as f:
    reader=csv.DictReader(f)
    rows=list(reader)
print("File loaded . Rows read :", len(rows))    
```

```python
columns=list(rows[0].keys())
print("Number of rows :", len(rows))
print("Number of columns :", len(columns))
print("Columns names :", columns)
```

```python
for row in rows[:5]:
    print(row)
```

```python
ids=[row["Customer_ID"]for row in rows]
print("Total IDs :" ,len(ids))
print("Unique IDs :" ,len(set(ids)))
print("Repeated IDs :", len(ids)-len(set(ids)))
```

```python
seen=set()
repeated=set()
for i in ids:
    if i in seen:
        repeated.add(i)
    else:
        seen.add(i)
print("IDs that appear more than once :", sorted(repeated))            
```

```python
row_tuples=[tuple(row.values()) for row in rows]
print("Total rows :", len(row_tuples))
print("Unique rows :", len(set(row_tuples)))
print("Exact duplicate rows :", len(row_tuples)-len(set(row_tuples)))
```

```python
seen=set()
repeated=set()
for t in row_tuples:
    if t in seen :
        print("Exact duplicate rows :", t)
    else:
        seen.add(t)
            
```
