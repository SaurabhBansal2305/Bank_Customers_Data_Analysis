```python
clean=bank.copy()
clean = clean.drop_duplicates().reset_index(drop=True)
valid = {
    "City": ["Indore", "Bhopal", "Delhi", "Mumbai", "Pune"],
    "Account_Type": ["Savings", "Salary", "Current"],
    "Employment_Type": ["Salaried", "Self_Employed", "Student", "Retired"],
    "Preferred_Channel": ["Mobile_App", "Net_Banking", "Branch", "ATM_Only"],
    }
for col, values in valid.items():
    lookup = {v.lower(): v for v in values}
    cleaned = clean[col].str.strip().str.lower().map(lookup)
    clean[col] = cleaned.fillna(clean[col])      
    print(col, clean[col].unique())           
```

```python
import numpy as np
ranges = {
    "Age": (18, 78),
    "Monthly_Income": (8000, 250000),
    "Tenure_Years": (0.2, 25),
    "Monthly_Transactions": (3, 70),
    "Avg_Monthly_Balance": (2000, None),
    }
for col, (lo, hi) in ranges.items():
    bad = pd.Series(False, index=clean.index)
    if lo is not None:
        bad |= clean[col] < lo
    if hi is not None:
        bad |= clean[col] > hi
    print(col, "out of range:", bad.sum())
    clean.loc[bad, col] = np.nan
```

```python
clean = clean.dropna(subset=["Customer_ID", "Avg_Monthly_Balance"])
for col in ["Age", "Tenure_Years", "Monthly_Transactions"]:
    clean[col] = clean[col].fillna(clean[col].median())
clean["Monthly_Income"] = clean["Monthly_Income"].fillna(
clean.groupby("Employment_Type")["Monthly_Income"].transform("median"))
clean["Monthly_Income"] = clean["Monthly_Income"].fillna(clean["Monthly_Income"].median())
for col in valid:
    clean[col] = clean[col].fillna("Unknown")
clean["Age"] = clean["Age"].round().astype(int)
clean["Monthly_Income"] = clean["Monthly_Income"].round().astype(int)
print("Missing left:", clean.isna().sum().sum())
print('\n',"Duplicate IDs:", clean["Customer_ID"].duplicated().sum())
print('\n',clean.shape)
    
```

```python
print("Missing values per column :") 
print(clean.isna().sum()[clean.isna().sum()>0])
print('\nDuplicate rows:', clean.duplicated().sum())
```

```python
print(clean.describe().round(1))
```

```python
clean["Income_per_Transaction"] = (clean["Monthly_Income"] / clean["Monthly_Transactions"]).round(2)

def segment(balance):
    if balance < 25000:
        return "Low Balance"
    elif balance < 100000:
        return "Healthy"
    else:
        return "Premium"

clean["Customer_Segment"] = clean["Avg_Monthly_Balance"].apply(segment)

clean["Tenure_Group"] = pd.cut(clean["Tenure_Years"],bins=[0, 2, 5, 10, 25],
    labels=["0-2 yrs", "2-5 yrs", "5-10 yrs", "10+ yrs"])

print(clean[["Income_per_Transaction", "Customer_Segment", "Tenure_Group"]].head())
print(clean["Customer_Segment"].value_counts())
print(clean["Tenure_Group"].value_counts())    
```
